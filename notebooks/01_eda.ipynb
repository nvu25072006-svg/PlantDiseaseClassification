# 01. Exploratory Data Analysis (EDA) & Stratified Split

```python
import sys
from pathlib import Path
import pandas as pd
from PIL import Image

sys.path.append(str(Path('..').resolve()))
from src.config import DATA_DIR, RESULTS_DIR, SEED, seed_everything
from src.data import scan_image_dataset, stratified_holdout_split
from src.evaluate import plot_class_distribution

seed_everything(SEED)
image_paths, labels, class_names = scan_image_dataset(DATA_DIR)
print(f'Total images: {len(image_paths)} | Classes: {len(class_names)}')
```

```python
df_eda = pd.DataFrame({'path': [str(p) for p in image_paths], 'label_id': labels})
df_eda['class_name'] = df_eda['label_id'].map(lambda idx: class_names[idx])
class_counts = df_eda['class_name'].value_counts()
display(class_counts.to_frame(name='count'))

plot_class_distribution(labels, class_names)
display(Image.open(RESULTS_DIR / 'fig3_1_class_distribution.png'))
if (RESULTS_DIR / 'fig3_2_feature_histograms.png').exists():
    display(Image.open(RESULTS_DIR / 'fig3_2_feature_histograms.png'))
if (RESULTS_DIR / 'fig3_3_sample_images_grid.png').exists():
    display(Image.open(RESULTS_DIR / 'fig3_3_sample_images_grid.png'))
```

```python
splits = stratified_holdout_split(image_paths, labels)
for split_name, (paths, y) in splits.items():
    print(f'{split_name.upper():5s}: {len(paths):4d} ({len(paths)/len(image_paths)*100:.1f}%)')
```
